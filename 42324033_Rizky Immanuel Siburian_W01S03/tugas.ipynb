```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
```

```python
# Tugas 1
matriks = np.random.rand(5, 5)
print("Matriks 5x5:")
print(matriks)
print("Nilai Maksimum:", np.max(matriks))
print("Nilai Minimum:", np.min(matriks))
print("Nilai Rata-rata:", np.mean(matriks))
```

```python
# Tugas 2
x_val = np.linspace(-1.5, 1.5, 1000)
y_val = np.linspace(-1.5, 1.5, 1000)
X2, Y2 = np.meshgrid(x_val, y_val)

Z2 = X2**6 + 3*X2**4*Y2**2 - 3*X2**4 + 3*X2**2*Y2**4 - X2**2*Y2**3 - 6*X2**2*Y2**2 + 3*X2**2 + Y2**6 - 3*Y2**4 + 3*Y2**2 - 1

plt.figure()
plt.contour(X2, Y2, Z2, levels=[0], colors='blue')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Grafik 2D Implicit Curve')
plt.grid(True)
plt.show()
```

```python
# Tugas 3
x_val3 = np.linspace(-3, 3, 300)
y_val3 = np.linspace(-3, 3, 300)
X3, Y3 = np.meshgrid(x_val3, y_val3)

Z3 = (5*Y3/4 - np.abs(X3))**2 + X3**2

fig3 = plt.figure()
ax3 = plt.axes(projection='3d')
ax3.plot_surface(X3, Y3, Z3, cmap='viridis')
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_zlabel('z')
ax3.set_title('Grafik 3D Surface')
plt.show()
```
