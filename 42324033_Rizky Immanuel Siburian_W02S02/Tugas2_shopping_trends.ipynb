```python
import pandas as pd
path = "shopping_trends.csv"  
df = pd.read_csv(path)
df
```

```python
prev_purchases = df['Previous Purchases']
```

```python
mean_val = prev_purchases.mean()
std_val = prev_purchases.std()
min_val = prev_purchases.min()
max_val = prev_purchases.max()
q1_val = prev_purchases.quantile(0.25)
q2_val = prev_purchases.quantile(0.50) 
q3_val = prev_purchases.quantile(0.75)
```

```python
print("=== STATISTIK PREVIOUS PURCHASES ===")
print(f"Mean (Rata-rata)   : {mean_val:.2f}")
print(f"Std (Standar Dev)  : {std_val:.2f}")
print(f"Min (Nilai Terendah): {min_val}")
print(f"Max (Nilai Tertinggi): {max_val}")
print(f"Q1 (Kuartil 1 / 25%): {q1_val:.2f}")
print(f"Q2 (Kuartil 2 / 50%): {q2_val:.2f}")
print(f"Q3 (Kuartil 3 / 75%): {q3_val:.2f}")
```

```python
corr_age_purchase = df[['Age', 'Purchase Amount (USD)']].corr()
```

```python
corr_purchase_review = df[['Purchase Amount (USD)', 'Review Rating']].corr()

print("=== KORELASI AGE vs PURCHASE AMOUNT (USD) ===")
print(corr_age_purchase)

print("\n=== KORELASI PURCHASE AMOUNT (USD) vs REVIEW RATING ===")
print(corr_purchase_review)
```
